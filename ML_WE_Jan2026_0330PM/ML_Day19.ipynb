### Reinforcement Leanring

A machine learns patterns from trial and error by getting reward and punishment

##### Main components Reinforcement Learning
- Agent =>   Who take decision
- Environment =>    Where Agent will work/perform
- Action =>         What Agent will do
- Reward =>         Feedback (positive/negative)
- State =>          Current Situation

Code Explanation
- Q Table -> memory where record and learn
- alpha -> fast learning
- gamma -> future reward's importance
- Loop -> to perform again n again

Example of a Reinforcement Learning
- An agent, who needs to go from 0 to 1
- Goal -> reach on stage 4 (reward = +10) (penalty -1)

```python
import numpy as np
import random
# total states 0 to 4
states = 5
actions = ['left','right']
# Q-table initialize
Q = np.zeros( (states , len(actions)) )
Q
```

```python
# Parameters
alpha = 0.8 # learning Rate
gamma = 0.9 # discount factor
episodes = 500
```

```python
# Reward function
def get_reward(state):
    if state==4:
        return 10
    else:
        return -1
```

```python
# Training
for episode in range(episodes):
    state = 0   # start from 0
    while state != 4:
        action = random.choice([0,1])
        # move
        if action == 0:   #left
            next_state = max(0,state-1)
        else:
            next_state = min(4,state+1)
        # Get Reward
        reward = get_reward(next_state)
        # Update Q-value
        Q[state,action] = Q[state,action]+alpha*( reward+gamma*np.max(Q[next_state])-Q[state,action] )
        state = next_state
```

```python
state = 0
print("Agent Path\n")
while state != 4:
    action = np.argmax(Q[state])
    print( f"State {state} -> {actions[action]}")
    if action == 0:
        state = max(0,state-1)
    else:
        state = min(4,state+1)
print("reached Goal!")
```
