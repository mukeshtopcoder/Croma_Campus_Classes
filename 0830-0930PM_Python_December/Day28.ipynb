```python
df.groupby('Country').agg({'Units Sold':'sum'})
```

```python
df['Country'] = df['Country'].map(lambda val: 'USA' if val=='United States of America' else val)
```

```python
qbc = df.groupby('Country').agg({'Units Sold':'sum'}).reset_index()
```

```python
qbc
```

```python
plt.figure(figsize=(6,3))
plt.plot( qbc['Country'] , qbc['Units Sold'] )
plt.show()
```

```python
plt.figure(figsize=(6,3))
plt.plot( qbc['Country'] , qbc['Units Sold'] )
plt.xlabel('Country')
plt.ylabel('Quantity')
plt.title('Sold Quantity By Country')
plt.show()
```

```python
plt.figure( figsize=(8,3) )
plt.bar(pbs['Segment'],pbs['Profit'])
plt.xlabel('Segment Names')
plt.ylabel("Profit Amount")
plt.title('Profit By Segment')
plt.show()
```

```python
plt.figure( figsize=(8,3) )
plt.bar(pbs['Segment'],pbs['Profit'] , color='lightgreen')
plt.xlabel('Segment Names')
plt.ylabel("Profit Amount")
plt.title('Profit By Segment')
plt.show()
```

```python
plt.figure( figsize=(8,3) )
colors = pbs['Profit'].map(lambda val : 'lightgreen' if val>0 else 'red')
plt.bar(pbs['Segment'],pbs['Profit'] , color=colors)
plt.xlabel('Segment Names')
plt.ylabel("Profit Amount")
plt.title('Profit By Segment')
plt.show()
```

```python
plt.figure(figsize=(6,3))
plt.plot( qbc['Country'] , qbc['Units Sold'] , marker='o' )
plt.xlabel('Country')
plt.ylabel('Quantity')
plt.title('Sold Quantity By Country')
plt.show()
```

```python
plt.figure(figsize=(6,3))
plt.plot( qbc['Country'] , qbc['Units Sold'] , marker='o' , mfc='red' , mec='yellow' )
plt.xlabel('Country')
plt.ylabel('Quantity')
plt.title('Sold Quantity By Country')
plt.show()
```

```python
plt.figure(figsize=(6,3))
plt.plot( qbc['Country'] , qbc['Units Sold'] , marker='o' , mfc='red' , mec='yellow' , linestyle='dotted' )
plt.xlabel('Country')
plt.ylabel('Quantity')
plt.title('Sold Quantity By Country')
plt.show()
```

```python
plt.figure(figsize=(6,3))
plt.plot( qbc['Country'] , qbc['Units Sold'] , marker='o' , mfc='red' , mec='yellow' , linestyle='dashed' )
plt.xlabel('Country')
plt.ylabel('Quantity')
plt.title('Sold Quantity By Country')
plt.show()
```

```python
plt.figure(figsize=(6,3))
plt.plot( qbc['Country'] , qbc['Units Sold'] , color='black', marker='o' , mfc='red' , mec='yellow' , linestyle='dashed' )
plt.xlabel('Country')
plt.ylabel('Quantity')
plt.title('Sold Quantity By Country')
plt.show()
```

```python
plt.figure(figsize=(6,3))
plt.plot( qbc['Country'] , qbc['Units Sold'] , color='black', marker='o' , mfc='red' , mec='yellow' , ls='-.' )
plt.xlabel('Country')
plt.ylabel('Quantity')
plt.title('Sold Quantity By Country')
plt.show()
```
