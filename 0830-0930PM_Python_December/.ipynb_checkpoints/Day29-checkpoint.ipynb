```python
df = sns.load_dataset('titanic')
```

```python
df.shape
```

```python
df.head(3)
```

```python
plt.figure(figsize=(8,3))
sns.heatmap(df.isnull()) 
plt.show()
```

```python
plt.figure(figsize=(8,3))
sns.histplot(df['age'] , kde=True)
plt.show()
```

```python
df = df.drop(columns=['deck'])
```

```python
df['age'] = df['age'].fillna(value = round(df['age'].mean(),0))
```

```python
plt.figure(figsize=(8,3))
sns.heatmap(df.isnull()) 
plt.show()
```

```python
plt.figure(figsize=(8,3))
sns.histplot(df['age'] , kde=True)
plt.show()
```

```python
df.head(2)
```

```python
plt.figure(figsize=(8,3))
sns.countplot(x='alive' , data=df)
plt.show()
```

```python
plt.figure(figsize=(8,3))
sns.countplot(x='alive' , hue='sex' , data=df)
plt.show()
```

```python
plt.figure(figsize=(8,3))
sns.countplot(x='alive' , hue='class' , data=df)
plt.show()
```

```python
plt.figure(figsize=(8,3))
sns.countplot(x='sex' , hue='class' , data=df)
plt.show()
```

```python
plt.figure(figsize=(8,3))
sns.boxplot(df['fare'])
plt.show()
```

```python
plt.figure(figsize=(8,3))
sns.distplot(df['fare'])
plt.show()
```

```python
plt.figure(figsize=(8,3))
sns.distplot(df[df['class']=='First']['fare'])
plt.show()
```
